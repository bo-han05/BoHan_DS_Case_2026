#### Pre-Modeling

```python
# Create target feature for easier training
train["kit_start"] = pd.to_datetime(train["kit_start"])
train["kit_end"] = pd.to_datetime(train["kit_end"])

train["build_duration_days"] = (
    train["kit_end"] - train["kit_start"]
).dt.total_seconds() / 86400
```

```python
print(train["build_duration_days"].describe())

print("Missing targets:", train["build_duration_days"].isna().sum() / len(train["kit_start"]), "%")
print("Negative timelines:", (train["build_duration_days"] < 0).sum() / len(train["kit_start"]), "%")
```

```python
# Remove rows with negative timelines as they make up a small percentage of the data
train = train[train["build_duration_days"] >= 0]
```

```python
# Normalize date features
train["download_date"] = pd.to_datetime(train["download_date"]).dt.normalize()
train["kit_start"] = pd.to_datetime(train["kit_start"]).dt.normalize()
train["kit_end"] = pd.to_datetime(train["kit_end"]).dt.normalize()
train["est_ship_date"] = pd.to_datetime(train["est_ship_date"]).dt.normalize()
calendar["date"] = pd.to_datetime(calendar["date"]).dt.normalize()
capacity["date"] = pd.to_datetime(capacity["date"]).dt.normalize()

# Join tables
df = train.merge(expected_minutes, on="family_desc", how="left", validate="m:1")
df = df.merge(calendar, left_on="kit_start", right_on = "date", how="left", validate="m:1").drop(columns=["date", "site_y"])
df = df.merge(capacity, left_on="kit_start", right_on = "date", how="left", validate="m:1").drop(columns=["date"])
```

```python
# Duplicate keys prevented merging with df
duplicates = capacity[capacity.duplicated("date", keep=False)].sort_values("date")
print(len(capacity), capacity["date"].nunique())
print(duplicates.head())
```

```python
capacity = capacity.drop_duplicates()

# Try join again
df = df.merge(capacity, left_on="kit_start", right_on = "date", how="left", validate="m:1").drop(columns=["date"])
```

```python
# Sanity check for df joining
print(df.isna().sum())
print(len(df))
```

```python
null_dates = df[df.fiscal_year.isna()]
print(null_dates[["sales_order_id", "download_date", "kit_start", "kit_end"]])
print(calendar.date.min(), calendar.date.max())
# kit_start dates start before min calendar date, number of rows is negligible so it is fine to drop

df = df.dropna(subset=["fiscal_year"])
```

```python
missing = df[df.expected_kit_minutes.isna()]
print(missing.family_desc.nunique(), "families missing of", df.family_desc.nunique())
print("\n", missing.family_desc.value_counts().head())
print("\n", missing.groupby("lob_desc").size().sort_values(ascending=False).head())
print("\n", missing.kit_start.dt.to_period("M").value_counts().sort_index())
# Take note of missing product order features, they account for 29% of the data, action can be taken later (imputation?)
```

```python
# Create numeric variables for XGBoost to train on
df["queue_days"] = (df["kit_start"] - df["download_date"]).dt.days
df["ship_slack_days"] = (df["est_ship_date"] - df["kit_start"]).dt.days
```

#### Modeling

```python
# Join features to inference
X_inf = inference.merge(expected_minutes, on="family_desc", how="left", validate="m:1")
X_inf = X_inf.merge(calendar, left_on="kit_start", right_on = "date", how="left", validate="m:1").drop(columns=["date", "site_y"])
X_inf = X_inf.merge(capacity, left_on="kit_start", right_on = "date", how="left", validate="m:1").drop(columns=["date"])
X_inf["queue_days"] = (X_inf["kit_start"] - X_inf["download_date"]).dt.days
X_inf["ship_slack_days"] = (X_inf["est_ship_date"] - X_inf["kit_start"]).dt.days

# X_inf.isna().sum()
```

```python
import xgboost as xgb # robust boosting model to use for initial baseline for case study

num_cols = ["buid", "tie_num", "quantity_produced", "line_qty", "expected_build_minutes", "expected_burn_minutes", "expected_kit_minutes"]
cat_cols = ["site_x", "mcid", "alt_type_desc", "lob_desc", "family_parent_desc", "family_desc", "fisc_week_val"]
drop_cols = ["download_date", "kit_start", "kit_end", "est_ship_date", "build_duration_days"]
features = [c for c in df.columns if c not in drop_cols]

# XGBoost needs numeric or category type
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
    X_inf[c] = pd.to_numeric(X_inf[c], errors="coerce")
for col in cat_cols:
    df[col] = df[col].astype('category')
    X_inf[col] = X_inf[col].astype(pd.CategoricalDtype(categories=df[col].cat.categories)) # maps unknown categories to NaN before prediction

X = df[features].copy()
y = df["build_duration_days"]
X_inf = X_inf[features].copy()   

model = xgb.XGBRegressor(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    tree_method="hist", enable_categorical=True,
    objective="reg:absoluteerror", random_state=42,
) # can tune hyperparameters and validate further down the road, past case study timeline
model.fit(X, y)

preds = np.rint(model.predict(X_inf))
pred_kit_end = inference["kit_start"] + pd.to_timedelta(preds, unit="D")
```

```python
results = pd.DataFrame({
    "Kit Start Date": inference["kit_start"],
    "Predicted Kit End Date": pred_kit_end
})

display(results)
# Can compare predicted vs. actual kit_end by site, order quantity, fiscal quarter, etc.
```

```python
feature_ranking = pd.DataFrame({
    'Feature': features,
    'Importance Score': model.feature_importances_
})

feature_ranking = feature_ranking.sort_values(by='Importance Score', ascending=False).reset_index(drop=True)

display(feature_ranking) # can look at SHAP values down the road as well
```
